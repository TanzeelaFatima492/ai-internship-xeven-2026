# PostgreSQL Using Python

## What is PostgreSQL?

PostgreSQL is an open-source Relational Database Management System (RDBMS). It stores data in tables and supports SQL (Structured Query Language) for managing and querying data.

### Features
- Open Source
- Secure
- Supports SQL
- High Performance
- ACID Compliant

## CRUD Operations

## EMBEDDING DIMENSION

The embedding dimension is the length of the vector.

#### Example:

Sentence : "I love pizza."

Embedding : [0.14, -0.23, 0.87, ..., 0.45]

If the vector contains: 384 numbers then the embedding dimension is 384.

Different models produce different dimensions.

### WHY DOES FAISS NEED THE EMBEDDING DIMENSION
FAISS stores vectors in an index.

When creating a FAISS index, we must specify
the vector dimension.

Example:

faiss.IndexFlatL2(384)

This tells FAISS that every vector added to the index
will contain exactly 384 values.

If you try to add a vector with a different size,
FAISS will raise an error.

Therefore, the embedding model and the FAISS index
must always use the same dimension.
