# Context Management

These are essential for building RAG (Retrieval Augmented Generation) systems and LangChain-based applications.

## 1. Context Window Limits

A context window is the maximum number of tokens an LLM can process at once.

### Example Token Limits:
- 4K tokens → small models  
- 8K tokens → standard models  
- 128K tokens → advanced models  

### Why it matters:
- Larger input = more memory usage
- Higher cost
- Slower response
- Hard limit for input size

## 2. Dynamic Context Injection

Instead of sending full data to the model, we retrieve only relevant information at runtime.

### Benefit:
- Saves tokens
- Improves accuracy
- Reduces noise

## 3. Memory Types in LLMs

LLMs use different memory systems:

### 1. Conversation Buffer Memory
Stores full chat history.



### 2. Summary Memory
Stores compressed summaries of old messages.



### 3. Entity Memory
Stores key facts like names, projects, and preferences.



## 4. Context Compression

When conversations become large, old data is compressed into summaries.

### Why?
- Saves tokens
- Keeps recent messages detailed
- Improves efficiency
