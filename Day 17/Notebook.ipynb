# Day 17: Embeddings and Semantic Similarity

Learn how AI converts text into numerical vectors and compares their meanings using mathematical techniques.

# What are Embeddings?

An embedding is a numerical vector that represents the semantic meaning of text.

Example:

dog → [0.24, 0.81, ..., 0.72]

Similar words such as "dog" and "puppy" have similar embeddings.

# Tokenization

Text is split into tokens.

"I love AI"

↓

["I", "love", "AI"]

↓

[51, 238, 1500]

Each token gets a unique Token ID.

# High-Dimensional Space

Embeddings are vectors in high-dimensional space.

2D:
(x,y)

3D:
(x,y,z)

AI:
(x1,x2,x3,...,x1536)

Similar meanings are located close together.


# Semantic Similarity

Semantic = Meaning

Similarity = Comparison

Semantic Similarity = Comparison based on meaning.

Examples:

Dog ↔ Puppy ✅

Car ↔ Automobile ✅

Dog ↔ Car ❌

# Similarity Metrics

## Cosine Similarity
Compares vector directions.

## Dot Product
Multiplies corresponding vector elements.

## Euclidean Distance
Calculates straight-line distance.

Cosine similarity is the most commonly used metric in AI.

# Key Learning

Computers process binary data, but AI converts text into embeddings that capture semantic meaning. Similarity metrics such as cosine similarity compare these vectors, while vector databases efficiently retrieve semantically related information. These concepts form the foundation of modern AI systems, including semantic search, recommendation engines, and RAG applications.
