# Python Calculator with Input, Type Conversion & Error Handling
This program takes two numbers from the user and performs basic arithmetic operations with proper error handling.

```python
try:
    first_number = int(input("Enter First number: "))
    second_number = int(input("Enter Second number: "))

    print("Adding of", first_number, "and", second_number)
    print("Result:", first_number + second_number)

    print("Subtraction of", first_number, "and", second_number)
    print("Result:", first_number - second_number)

    print("Multiplication of", first_number, "and", second_number)
    print("Result:", first_number * second_number)

    print("Quotient of", first_number, "and", second_number)
    print("Result:", first_number / second_number)

except ValueError:
    print("Error: Please enter valid integers only (no text allowed).")

except ZeroDivisionError:
    print("Error: Division by zero is not allowed.")
```

## How it works

- `input()` takes data as string
- `float()` converts it into number
- Arithmetic operations are performed on numbers
- `f-strings` format output clearly
- `try-except` handles errors safely:
  - ValueError → when input is not a number
  - ZeroDivisionError → when dividing by zero
