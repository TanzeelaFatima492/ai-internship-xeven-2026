# What is Machine Learning?

Machine Learning (ML) is a branch of Artificial Intelligence (AI) that enables computers to learn from data and make predictions or decisions without being explicitly programmed.

## Real-World Applications
- Email spam detection
- House price prediction
- Recommendation systems
- Medical diagnosis
- Self-driving cars

# Types of Machine Learning

Machine Learning has three main types:

1. Supervised Learning
2. Unsupervised Learning
3. Reinforcement Learning

# Supervised Learning

Supervised learning uses labeled data, where each input has a correct output.

## How It Works
1. Input data and labels are provided.
2. The model learns patterns.
3. It predicts outputs for new data.

## Examples
- Email spam detection
- House price prediction
- Image classification

# Unsupervised Learning

Unsupervised learning works with unlabeled data and discovers hidden patterns.

## Examples
- Customer segmentation
- Fraud detection
- Data clustering


# Reinforcement Learning

Reinforcement learning allows an agent to learn through trial and error using rewards and penalties.

## Components
- Agent
- Environment
- Actions
- Rewards

## Examples
- Robotics
- Game AI
- Recommendation systems

# Regression vs Classification

Regression predicts continuous values.

Examples:
- House prices
- Temperature
- Sales

Classification predicts categories.

Examples:
- Spam or Not Spam
- Healthy or Sick
- Cat or Dog

| Regression | Classification |
|-----------|---------------|
| Numerical values | Categories |
| Continuous output | Discrete output |
| House prices | Spam detection |
| Temperature | Disease detection |
