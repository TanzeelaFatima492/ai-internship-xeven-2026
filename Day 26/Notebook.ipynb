# What Are AI Agents?

## Definition
An **AI Agent** is a large language model (LLM) that can:
- **Use tools** (calculator, web search, databases, APIs)
- **Make decisions** autonomously
- **Plan and execute** multi-step tasks without human intervention

## How They Work
Instead of just generating text, agents:
1. Understand the user's goal
2. Decide which tool to use
3. Call the tool with appropriate inputs
4. Process the tool's output
5. Decide the next step (more tools or final answer)
6. Return the final result to the user

## Real-World Analogy
Think of an agent like a **smart assistant**:
- You ask: "What's the weather in Lahore and should I carry an umbrella?"
- It checks weather API → gets forecast → analyzes rain probability → advises you

## Key Difference from Regular LLMs
| Regular LLM | Agent |
|-------------|-------|
| Only generates text from training data | Can take real-world actions |
| Static knowledge (training cutoff) | Can access live information |
| No external tool usage | Uses calculators, APIs, databases |
| Single response | Multi-step reasoning + acting |

# The ReAct Framework

## What is ReAct?
**ReAct** = **Re**asoning + **Act**ing

A framework where the agent alternates between:
- **Thinking** about what to do
- **Acting** by calling a tool
- **Observing** the result
- **Repeating** until it can answer

## The ReAct Loop



### Step-by-Step Breakdown:

| Step | Name | What Happens | Example |
|------|------|--------------|---------|
| 1 | **Question** | User asks something | "What is 1970 divided by 4?" |
| 2 | **Thought** | Agent thinks about what to do | "I need to use the calculator" |
| 3 | **Action** | Agent calls a tool | `calculator(1970/4)` |
| 4 | **Observation** | Agent receives result | "492.5" |
| 5 | **Thought** | Agent decides if done | "I have the answer" |
| 6 | **Final Answer** | Returns to user | "1970 divided by 4 equals 492.5" |

## Why ReAct Works
- **Reduces hallucination** : agent relies on real tool outputs, not guesses
- **Transparent reasoning** : you can see what the agent is thinking
- **Debuggable** : if something fails, you know which step went wrong

# Types of AI Agents

## 1. Zero-Shot Agent
- **Description:** Uses tool descriptions alone to decide which tool to call
- **How:** LLM reads tool names + descriptions, infers which one fits the query
- **Example:** "What's 5+3?" → LLM reads calculator description → calls it
- **Pros:** Simple, no examples needed
- **Cons:** May pick wrong tool for ambiguous queries

## 2. Conversational Agent
- **Description:** Agent that maintains conversation history
- **How:** Uses memory buffer to remember previous Q&A pairs
- **Example:**
  - User: "What is RAG?"
  - Agent: "RAG is Retrieval-Augmented Generation..."
  - User: "What are its benefits?" (agent remembers context)
  - Agent: "RAG reduces hallucinations, improves accuracy..."
- **Pros:** Natural multi-turn conversations
- **Cons:** Memory grows, may need pruning/summarization

## 3. Structured Tool Agent
- **Description:** Agent that expects tools with strict input/output schemas
- **How:** Tools define exact parameter types (string, int, object)
- **Example:** `get_weather(city: str, date: str) → dict`
- **Pros:** Reliable, validated inputs
- **Cons:** Less flexible, needs well-defined schemas

## 4. OpenAI Functions Agent
- **Description:** Agent that uses OpenAI's function-calling format
- **How:** Tools are described as JSON schemas; LLM outputs function calls natively
- **Example:**
  ```json
  {
    "name": "get_weather",
    "arguments": {"city": "Lahore", "unit": "celsius"}
  }


## Types of Tools Agents Use

| Category | Examples |
|----------|---------|
| **Search** | Web search, document search (RAG) |
| **Computation** | Calculator, unit converter |
| **Data Access** | Database queries, API calls |
| **Time** | Current date/time, date calculations |
| **System** | File operations, email sending |

# The Agent Execution Loop

## What is the Agent Loop?
The **agent loop** is the continuous cycle an agent runs to solve a problem. It keeps iterating until:
- It has a final answer
- It reaches maximum iterations (safety limit)
- The user stops it

# Challenges in AI Agents

## 1. Tool Selection Errors
**Problem:** Agent picks the wrong tool for a task

| Error Type | Example |
|------------|---------|
| Wrong tool | Uses calculator for "What is RAG?" |
| Missed tool | Should use web search but tries to answer from memory |
| Confused tools | Uses RAG search instead of web search |

**Solutions:**
- Write clear, distinct tool descriptions
- Use structured tool formats (JSON schemas)
- Add examples in the prompt (few-shot)


**Solutions:**
- Set `max_iterations` limit (e.g., 5 or 10)
- Implement `early_stopping` when no progress
- Detect repeated identical actions
- Force a final answer after N attempts

## 3. Cost Management
**Problem:** Each LLM call costs tokens/money; agents make multiple calls

**Cost Factors:**
| Factor | Impact |
|--------|--------|
| LLM calls per step | Each Thought → Action → Observation uses tokens |
| Tool calls | Web search APIs may have their own costs |
| Long prompts | Tool descriptions + history = more input tokens |
| Iterations | More steps = higher cost |

**Solutions:**
- Use smaller/cheaper models for simple tasks
- Limit `max_iterations`
- Cache frequent web search results
- Summarize conversation history instead of full buffer
- Track and log all API costs

## Summary Table

| Challenge | Cause | Prevention |
|-----------|-------|------------|
| Wrong tool selection | Ambiguous tool descriptions | Clear, specific descriptions + examples |
| Infinite loops | No progress between steps | Max iterations, repeated action detection |
| High costs | Many LLM + API calls | Limit steps, cache results, use cheaper models |
| Hallucination | Agent ignores tool output | Force Final Answer from Observation only |
| Parsing errors | LLM outputs wrong format | Handle parsing errors gracefully, retry |

# Regular LLM vs AI Agent: Comparison

| Feature | Regular LLM | AI Agent |
|---------|-------------|----------|
| **Knowledge** | Fixed (training cutoff) | Live (can search web, query databases) |
| **Accuracy** | May hallucinate | Tool-grounded responses |
| **Math** | Often incorrect | Uses calculator for precision |
| **Real-time data** | Not available | Web search, APIs |
| **Complex tasks** | Single response | Multi-step planning |
| **Transparency** | Black box | Shows reasoning steps |
| **Cost** | 1 API call per query | Multiple calls (LLM + tools) |
| **Latency** | Fast (single response) | Slower (multiple iterations) |

## When to Use an Agent
- User needs **real-time information** (weather, news, stock prices)
- Task requires **multiple steps** (research, calculations, comparisons)
- Answer must be **factual and verifiable** (citing sources)
- Task involves **external systems** (databases, APIs, file systems)

## When a Regular LLM is Better
- Simple Q&A from static knowledge
- Creative writing, brainstorming
- Low-latency requirements
- Cost-sensitive applications
