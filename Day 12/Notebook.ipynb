# Python Functions - Basic Concepts
This notebook explains functions, arguments, scope, and design principles in Python.

## 1. Function Basics

A function is a reusable block of code that performs a task.

- Defined using `def`
- Can take inputs (parameters)
- Can return output using `return`
- If no return → returns `None`

```python
def greet(name):
    return "Hello " + name

print(greet("Ali"))
```

## 2. Types of Arguments

- Positional arguments → order matters  
- Keyword arguments → use parameter names  
- Default arguments → fallback values  
- *args → multiple positional values  
- **kwargs → multiple keyword values  

```python
def show(name, age=18):
    print(name, age)

show("Ali")
show(age=20, name="Sara")
```

## 3. Scope

- Local variable : inside function  
- Global variable : outside function  
- `global` keyword : modify global variable  

## 3. Scope

- Local variable → inside function  
- Global variable → outside function

## 4. Function Design Principles

- Single responsibility (one task per function)
- Use descriptive names
- Add docstrings for explanation
