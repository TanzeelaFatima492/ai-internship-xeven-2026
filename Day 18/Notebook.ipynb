## Key Terminologies

### Chunking
Process of breaking large documents into smaller parts for embeddings and retrieval.

### Embeddings
Vector representation of text used for semantic similarity search.

### RAG (Retrieval Augmented Generation)
A system that retrieves relevant document chunks and sends them to an LLM for answering questions.

### Token
Smallest unit of text processed by language models.

### Context Window
Maximum number of tokens an LLM can process at once.

### Overlap
Shared portion between consecutive chunks to preserve context.

## Chunk Size Comparison Results

### Chunk Size: 200 characters
- Number of chunks: Very High
- Context preservation: Low
- Precision: Very High
- Retrieval quality: Medium–High
- Issue: Information fragmentation

### Chunk Size: 500 characters
- Number of chunks: Medium
- Context preservation: Balanced
- Precision: High
- Retrieval quality: Best
- Recommendation: Optimal for RAG systems

### Chunk Size: 1000 characters
- Number of chunks: Low
- Context preservation: High
- Precision: Medium
- Retrieval quality: Good
- Issue: Some irrelevant context included

### Chunk Size: 2000 characters
- Number of chunks: Very Low
- Context preservation: Very High
- Precision: Low
- Retrieval quality: Weak
- Issue: Too much noise in single chunk

## LangChain Splitter Concepts

### RecursiveCharacterTextSplitter
- Splits using paragraphs → sentences → characters
- Best general-purpose splitter

### TokenTextSplitter
- Splits based on token count
- Best for embedding models

### MarkdownHeaderTextSplitter
- Splits based on headings (#, ##, ###)
- Best for structured documents

## Metadata Per Chunk

Each chunk contains:

- Source file name
- Chunk ID
- Chunk size
- Document type
- Section header
- Token estimate


### Example Structure:

Chunk:
"Machine Learning is a subset of AI"

Metadata:
- Source: learnings.md
- Type: text
- Section: AI Basics
- Chunk ID: 3
- Tokens: ~8

## Trade-offs Analysis

### Small Chunks (200–500)
High precision  
Better search results  
Low context  
More chunks required  

### Medium Chunks (500–1000)
Balanced approach  
Best retrieval quality  
Recommended for RAG  

### Large Chunks (1000–2000)
High context  
Better summarization  
Low precision  
Noisy retrieval results  

## Key Insight:
Smaller chunks improve precision but lose context.
Larger chunks improve context but reduce accuracy.
Medium chunks provide the best balance.
