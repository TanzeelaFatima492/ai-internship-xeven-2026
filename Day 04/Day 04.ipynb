```python
a = 10
b = 3

print("Addition:", a + b)
print("Subtraction:", a - b)
print("Multiplication:", a * b)
print("Division:", a / b)
print("Floor Division:", a // b)
print("Modulus:", a % b)
print("Exponent:", a ** b)
```

```python
x = 15
y = 20

print(x == y)
print(x != y)
print(x > y)
print(x < y)
print(x >= y)
print(x <= y)
```

```python
age = 20
has_id = True

print(age > 18 and has_id)
print(age > 18 or has_id)
print(not has_id)
```

```python
print(int("10"))
print(float("5.5"))
print(str(100))
print(bool(1))
print(bool(0))
```
