# 🔹 Functions in Python

A function is a reusable block of code used to perform a specific task.

### Types of Arguments:
- Positional arguments
- Keyword arguments
- Default arguments
- *args → variable positional arguments
- **kwargs → variable keyword arguments

```python
def greet(name):
    return "Hello " + name

print(greet("Ali"))
```

# 🔹 *args and **kwargs

- `*args` → takes multiple positional arguments (tuple)
- `**kwargs` → takes multiple keyword arguments (dictionary)

```python
def demo(*args, **kwargs):
    print("args:", args)
    print("kwargs:", kwargs)

demo(1, 2, 3, name="Ali", age=20)
```

# 🔹 Lambda Functions

A lambda function is a small anonymous function written in one line.

Used for simple operations.

```python
square = lambda x: x * x
print(square(5))
```

# 🔹 List Comprehension

A short way to create lists in a single line using loops and conditions.

```python
nums = [1, 2, 3, 4, 5]

squares = [x * x for x in nums]
evens = [x for x in nums if x % 2 == 0]

print(squares)
print(evens)
```

# 🔹 Dictionary Comprehension

A short way to create dictionaries using loops in one line.

```python
nums = [1, 2, 3, 4]

squares = {x: x * x for x in nums}
even_squares = {x: x * x for x in nums if x % 2 == 0}

print(squares)
print(even_squares)
```
