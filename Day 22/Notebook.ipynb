This notebook explains the complete implementation of a FAISS-based document retrieval system.

The system performs:

- Text chunking
- Embedding generation using SentenceTransformer
- Vector storage using FAISS
- Similarity search with metadata filtering

The project demonstrates how vector databases enable semantic search compared to traditional keyword-based search.

WHAT IS A VECTOR DATABASE

A vector database stores high-dimensional vector representations of text, images, or other data types.

Each document is converted into embeddings using a neural model.
These embeddings represent semantic meaning instead of raw text.

Key idea:
Similar meaning → closer vectors in space
Different meaning → far apart vectors

WHY NOT TRADITIONAL DATABASES

Traditional databases (SQL / NoSQL) are not suitable for semantic search.

Limitations:
- Work on exact matching
- Cannot understand meaning
- No support for similarity search

Vector databases solve this using Approximate Nearest Neighbor (ANN) search algorithms.

INDEXING ALGORITHMS USED IN VECTOR DATABASES

1. HNSW (Hierarchical Navigable Small World)
- Graph-based indexing
- Very fast search
- High accuracy

2. IVF (Inverted File Index)
- Clustering-based approach
- Faster on large datasets
- Slight trade-off in accuracy

FAISS supports both methods for scalable vector search.

KEY OPERATIONS IN VECTOR DATABASES

1. Add Vectors
- Store embeddings in index

2. Similarity Search
- Retrieve K nearest neighbors using distance metrics

3. Metadata Filtering
- Filter results using attributes like source, category, or section

PERFORMANCE METRICS

1. Recall
- Measures accuracy of retrieved results
- Higher recall means better semantic matching

2. Latency
- Time taken to return results
- Lower latency means faster search

3. Throughput
- Number of queries handled per second
