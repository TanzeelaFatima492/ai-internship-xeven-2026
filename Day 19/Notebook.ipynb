# Prompt Engineering

##  Core Prompting Techniques

### 1. Zero-shot Prompting
- No examples provided
- Only instruction
- Fast but less accurate

### 2. Few-shot Prompting
- 2–5 examples
- Improves pattern learning
- Higher accuracy

### 3. Chain-of-Thought (CoT)
- Step-by-step reasoning
- Best for logic/maths/complex tasks

##  Prompt Structure (Prompt Patterns)

- Persona → Role of AI
- Instruction → Task definition
- Context → Background info
- Format → Output structure
- Examples → Demonstrations
- Constraints → Rules/limits

##  System Message
Defines global behavior of AI:
- Tone
- Role
- Restrictions
- Output format rules

## 📌 Prompt Template Library

We design reusable templates for:

- Summarization
- Extraction
- Generation
- Analysis

Each template contains:
- System message
- User instruction
- Variables like {text}, {format}, {constraints}
- Optional examples

This allows reusable and scalable prompt engineering.
