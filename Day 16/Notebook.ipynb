## 🔗 2. LCEL & Pipeline Mechanics

### ✦ LCEL (LangChain Expression Language)
LangChain ka apna makhsoos declarative syntax jiske zariye hum alag-alag components (Prompts, Models, Parsers) ko ek seedhi line mein jor kar production pipelines banate hain.

### ✦ The Pipe Operator (`|`)
Python mein isay standard tor par *Bitwise OR* kehte hain, lekin LangChain framework ke andar iska kaam bilkul badal diya gaya hai:
> **Rule:** Pehle component ka jo bhi output aayega, use automatic agle component ka input bana do.



### ✦ Core Pipeline Architecture Functions
* **Runnable Interface:** LangChain ka ek standard rule/protocol. Jo bhi component is rule par banta hai, uske paas `.invoke()` aur `.stream()` ke functions laazmi hote hain taaki woh pipe (`|`) ke sath connect ho sakay.
* **`.invoke()`:** Iska matlab hai **"Run/Execute"**. Jab hum poori chain ko input variables ka data de kar `.invoke()` chalate hain, toh hamari pipeline active ho jati hai.
* **StrOutputParser:** Model ka jo response aata hai, usme bohot saara faltu data (meta-data, token counts, JSON tags) hota hai. Yeh parser us pure kachre ko saaf karke hamen **sirf clean text string** nikaal kar deta hai.


## 📄 3. Data Ingestion & Context Engineering



### ✦ Document Loaders
Yeh woh built-in functions ya tools hain jo computer ki bikhri hui files (PDFs, Web links, CSV sheets) ko read karke LangChain ke parhne ke kabil banate hain (e.g., `TextLoader`, `PyPDFLoader`).

### ✦ The Document Object
LangChain jab bhi kisi file ko load karta hai, toh usay ek standard structural container mein pack karta hai jise **Document Object** kehte hain. Iske hamesha do compartments hote hain:
1. **`page_content`**: File ke andar ka asal pure text data.
2. **`metadata`**: File ki extra jankari ki dictionary (jaise: file ka path kya hai, website ka URL kya hai, ya text kis page number se uthaya gaya).

### ✦ Boundary Control Terms
* **Context Window / Limit:** Ek Large Language Model (LLM) ek single waqt mein kitne words ya tokens parh aur yaad rakh sakta hai, uski aakhri had (boundary) ko Context Window kehte hain.
* **Truncate:** Agar hamari file bohot badi ho aur model ki limit cross kar rahi ho, toh code crash hone se bachane ke liye text ko aakhir se kaat kar chota karne ke amal ko truncate kehte hain (e.g., `text[:4000]`).


## 💻 4. Python Backend & Environment Terms

* **Environment Variables (`.env`):** Ek makhsoos local file jahan hum apni sensitive API keys chupa kar rakhte hain, taaki code GitHub par upload karte waqt hamari security leak na ho.
* **python-dotenv:** Woh helper library jo `.env` file se secure keys utha kar hamare chalte hue Python program ki memory mein system variables ki tarah load karti hai.
* **`--break-system-packages`:** Python ke naye versions (jaise Python 3.14) mein global environment mein pip install karne ke liye yeh flag lagana zaroori hota hai. Yeh operating system ko overwrite karne ki ijazat deta hai.
* **`__pycache__` (Cache Files):** Python jab kisi script ko chalata hai, toh backup par machine code files banata hai. Agar hum apni python file ka naam badlein (Rename karein), toh yeh puraani cache system ko confuse karke `ImportError` paida karti hai jab tak isay manually delete na kiya jaye.
