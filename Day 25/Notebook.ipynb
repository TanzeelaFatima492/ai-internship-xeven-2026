# RAG Advanced Retrieval Strategies

## 1. Dense, Sparse, Hybrid Retrieval

### Dense Retrieval (Embeddings)
- Meaning-based search using vector embeddings  
- Finds semantic similarity (not exact words)  
- Example: “car” ≈ “automobile”  

### Sparse Retrieval (BM25)
- Keyword-based traditional search  
- Matches exact terms in query and documents  
- Example: “car repair” → documents with same words rank higher  

### Hybrid Retrieval
- Combination of Dense + Sparse  
- Balances semantic + keyword matching  
- Used in modern search systems for best accuracy  

## 2. Re-ranking (Two-Stage Retrieval)

### Process:
- Step 1: Retrieve many documents (e.g., 50–100)
- Step 2: Select top candidates (e.g., 20)
- Step 3: Re-rank using LLM / cross-encoder
- Step 4: Final selection (top 5)

### Benefit:
- Improves relevance and quality of final answers

## 3. Query Expansion

### Idea:
Generate multiple versions of the same query

### Example:
User Query: “AI job”

Expanded Queries:
- “artificial intelligence jobs”
- “machine learning engineer roles”
- “AI careers opportunities”

### Benefit:
- Increases recall (more relevant documents found)

## 4. Self-Query Retrieval

### Idea:
LLM converts natural language into structured filters

### Example:
User Query:
“AI jobs in Pakistan after 2020”

Structured Output:
- Topic: AI jobs  
- Location: Pakistan  
- Year filter: > 2020  

### Benefit:
- Enables precise filtering + better retrieval accuracy

## 5. Parent Document Retrieval

### Idea:
- Retrieve small relevant chunks first  
- Then return full original document (parent)

### Flow:
Chunk → Match → Retrieve → Expand to full document

### Benefit:
- Maintains full context instead of partial text only  

## Final Comparison Table

| Strategy | Purpose | Key Benefit |
|----------|--------|-------------|
| Dense | Semantic search | Meaning-based matching |
| Sparse | Keyword search | Exact match precision |
| Hybrid | Combined search | Best overall retrieval |
| Re-ranking | Result filtering | Higher quality outputs |
| Query Expansion | Query variation | Better recall |
| Self-query | Structured filtering | Precise retrieval |
| Parent retrieval | Context expansion | Full document context |
