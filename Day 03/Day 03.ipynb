# 1. Boolean Logic in Programming

Boolean logic is the foundation of decision-making in programming.

It has only two values:
- True
- False

## Main Boolean Operators

- AND → both conditions must be True
- OR → at least one condition must be True
- NOT → reverses condition

# 2. Writing Clean Conditional Statements in Python

Good conditional statements should be:
- Easy to read
- Well structured
- Minimal nesting
- Logically ordered
- Handle edge cases

# 3. Best Practices for if-elif-else

- Use elif instead of multiple if statements
- Order conditions from specific to general
- Always handle edge cases
- Avoid deep nesting
- Use meaningful variable names

## Task 1 :

```python
"""
Age Verification System
Classifies user into: Child, Teenager, Adult, Senior
Handles: negative age, non-numeric input, empty name
"""

name = input("Enter your name : ")

# Check if name is empty
if name.strip() == "":
    print("Error: Name cannot be empty.")
else:
    try:
        age = int(input("Enter your age  : "))

        # Invalid negative age check (IMPORTANT: must be first)
        if age < 0:
            print(f"\nSorry {name}, age cannot be negative.")

        # Extra validation for unrealistic age
        elif age > 150:
            print(f"\nSorry {name}, that age is unrealistic.")

        # Age categories
        elif age < 13:
            print(f"\nHello {name}! As a child, enjoy your childhood — the world is full of wonder!")

        elif age <= 17:
            print(f"\nHello {name}! As a teenager, you have many opportunities ahead.")

        elif age <= 64:
            print(f"\nHello {name}! As an adult, you are in your prime — keep chasing your goals!")

        else:
            print(f"\nHello {name}! As a senior, your wisdom and experience are truly valuable.")

    except ValueError:
        print(f"\nSorry {name}, please enter a valid number for age.")
```

```python
"""
Grade Calculator with Feedback
Assigns letter grade A-F with personalized message
Handles: out-of-range, non-numeric, empty name
"""

student_name = input("Enter student name         : ")

if student_name.strip() == "":
    print("Error: Name cannot be empty.")
else:
    try:
        grade = float(input("Enter numeric grade (0-100): "))

        if grade < 0:
            print(f"\n{student_name}, grade cannot be below 0.")
        elif grade > 100:
            print(f"\n{student_name}, grade cannot exceed 100.")
        elif grade >= 90:
            letter, feedback, desc = "A", "Excellent work!", "Outstanding — you mastered the material."
        elif grade >= 80:
            letter, feedback, desc = "B", "Good job!", "Above average — strong understanding."
        elif grade >= 70:
            letter, feedback, desc = "C", "Keep it up!", "Average — a bit more effort needed."
        elif grade >= 60:
            letter, feedback, desc = "D", "You can do better!", "Below average — review and practice."
        else:
            letter, feedback, desc = "F", "Don't give up!", "Needs improvement — seek help."

        if 0 <= grade <= 100:
           print("Invalid grade entered. Grade must be between 0 and 100.")

    except ValueError:
        print(f"\n{student_name}, please enter a numeric grade (e.g. 85).")
```
