# Python Data Structures Comparison
## List vs Tuple vs Set vs Dictionary

This notebook explains the four major Python data structures with:
- Properties (mutability, ordering)
- Time complexity (access, search, insert, delete)
- Practical code examples
# 📌 Comparison Table

| Feature        | List                | Tuple              | Set                | Dictionary (Dict)     |
|----------------|---------------------|--------------------|--------------------|----------------------|
| Mutability     | Mutable             | Immutable          | Mutable            | Mutable              |
| Ordering       | Ordered             | Ordered            | Unordered          | Ordered (3.7+)       |
| Duplicates     | Allowed             | Allowed            | Not allowed        | Keys unique          |
| Indexing       | Yes                 | Yes                | No                 | Key-based            |
| Syntax         | [1,2,3]             | (1,2,3)            | {1,2,3}            | {"a":1,"b":2}        |

# ⏱ Time Complexity Overview

| Operation     | List        | Tuple       | Set         | Dictionary |
|----------------|------------|-------------|-------------|------------|
| Access         | O(1)       | O(1)        | O(1)*       | O(1)       |
| Search         | O(n)       | O(n)        | O(1) avg    | O(1) avg   |
| Insert         | O(1)/O(n)  | Not allowed | O(1) avg    | O(1) avg   |
| Delete         | O(n)       | Not allowed | O(1) avg    | O(1) avg   |

 Set access is membership-based, not index-based.

```python
# LIST: ordered, mutable, allows duplicates

my_list = [10, 20, 30, 20]

my_list.append(40)      # Insert
my_list.remove(20)      # Delete first occurrence

print("List:", my_list)

# Access
print("First element:", my_list[0])
```

```python
# TUPLE: ordered, immutable

my_tuple = (10, 20, 30)

print("Tuple:", my_tuple)

# Access
print("First element:", my_tuple[0])

# my_tuple[0] = 100  ❌ Not allowed (immutable)
```

```python
# SET: unordered, unique values

my_set = {10, 20, 30, 20}

my_set.add(40)        # Insert
my_set.remove(10)     # Delete

print("Set:", my_set)

# Membership check (fast)
print(20 in my_set)
```

```python
# DICTIONARY: key-value pairs, fast lookup

my_dict = {
    "name": "Tanzeela",
    "age": 20,
    "course": "IT"
}

# Insert / Update
my_dict["city"] = "Kotli"

# Access
print("Name:", my_dict["name"])

# Delete
del my_dict["age"]

print("Dictionary:", my_dict)
```

# 🧠 When to Use Which?

### ✔ List
- When order matters
- When you need frequent changes (add/remove)

### ✔ Tuple
- When data should NOT change
- Faster than list (read-only data)

### ✔ Set
- When uniqueness is important
- Fast membership checking

### ✔ Dictionary
- When you need key-value mapping
- Fast lookup by key (like database records)
