## What is SQLAlchemy?

SQLAlchemy is a Python library (ORM - Object Relational Mapper) that allows Python programs to communicate with relational databases such as PostgreSQL, MySQL, SQLite, and Oracle.

Instead of writing raw SQL queries like:

SELECT * FROM users;

we can write Python code:

users = session.query(User).all()

SQLAlchemy converts Python code into SQL queries automatically.

### Why SQLAlchemy?

Advantages:
- Less SQL writing
- Cleaner code
- Database independent
- Easier maintenance
- Supports relationships
- Works well with FastAPI

### ORM (Object Relational Mapper)

ORM stands for Object Relational Mapper.

Object = Python Class

Relation = Database Table

Mapper = Converts Python objects into database records.

Database Table

users

| id | name | email |
|----|------|-------|
| 1  | John | john@gmail.com |

SQLAlchemy maps the Python class to the database table automatically.

### SQL vs SQLAlchemy

Without SQLAlchemy

INSERT INTO users(name,email)
VALUES('John','john@gmail.com');

With SQLAlchemy

user = User(
    name="John",
    email="john@gmail.com"
)

session.add(user)
session.commit()

Notice:
We never write SQL ourselves.
SQLAlchemy generates it automatically.

### Pydantic Schema:

A Pydantic schema defines the structure of data entering or leaving a FastAPI application. It validates user input automatically and ensures only correctly formatted data is processed. Schemas are also used to control which fields are returned in API responses.

### EmailStr:

EmailStr is a special Pydantic type that validates whether a string is a properly formatted email address. If the email is invalid, FastAPI automatically returns a validation error before the request reaches our application logic.

### Difference between SQLAlchemy Models and Pydantic Schemas

| SQLAlchemy Model            | Pydantic Schema                      |
| --------------------------- | ------------------------------------ |
| Represents a database table | Represents API request/response data |
| Stores data in PostgreSQL   | Validates incoming and outgoing data |
| Used by SQLAlchemy ORM      | Used by FastAPI                      |


## Hashing:

Hashing is a one-way mathematical process. Password hashing converts a plain-text password into a fixed-length string using a one-way algorithm. Unlike encryption, hashing cannot be reversed, making it a secure way to store passwords.

### Encryption vs Hashing

| Encryption                         | Hashing                 |
| ---------------------------------- | ----------------------- |
| Can be decrypted                   | Cannot be reversed      |
| Uses encryption & decryption keys  | Uses one-way algorithms |
| Used for sensitive data like files | Used for passwords      |


### bcrypt

bcrypt is a password hashing algorithm that automatically generates a unique salt for every password and is intentionally slow. This makes it much more resistant to brute-force and rainbow table attacks than general-purpose hashing algorithms.

### APIRouter

APIRouter helps organize related API endpoints into separate files. Instead of writing all routes inside main.py, FastAPI allows us to group related functionality, such as authentication or chat, into dedicated router modules.  

`include_router()` Registers routers with the FastAPI application

### Dependency Injection

Depends(get_db) automatically provides a database session to the endpoint. FastAPI creates the session before the request and closes it afterward, preventing connection leaks and reducing boilerplate code.

### JWT

JWT stands for JSON Web Token.

Think of it like a digital ID card.

JWT is a compact, URL-safe token used to securely transmit user identity between a client and a server. After successful authentication, the server generates a signed token that the client includes in future requests instead of sending the username and password again.

JWT enables stateless authentication. The server does not need to store user sessions because all required authentication information is contained within the signed token.

### Secret Key

The secret key is used to sign JWTs. Only the server should know this key. If the key changes, previously issued tokens become invalid.

### Authentication

Authentication is the process of verifying a user's identity. During login, the application checks whether the provided email exists and whether the entered password matches the securely stored hashed password.
